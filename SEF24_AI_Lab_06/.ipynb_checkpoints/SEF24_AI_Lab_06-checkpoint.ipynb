## Step 2: Implement Inference Rules

You will implement core logical inference rules.

These rules help derive new knowledge from existing facts.

```python
class InferenceEngine:
    def modus_ponens(self, kb):
        # Find P and (P => Q), infer Q
        inferred = set()
        facts = set()
        implications = []
        for s in kb.sentences:
            if '=>' in s and '<=>' not in s:
                left, right = s.split('=>')
                implications.append((left.strip(), right.strip()))
            elif 'AND' not in s and 'OR' not in s and 'NOT' not in s:
                facts.add(s.strip())
        for left, right in implications:
            if left in facts and right not in facts:
                inferred.add(right)
        for i in inferred:
            if i not in kb.sentences:
                kb.add(i)
        return inferred

    def and_elimination(self, sentence):
        # From (A AND B), infer A and B
        if 'AND' in sentence:
            parts = [p.strip() for p in sentence.split('AND')]
            return set(parts)
        return set()

    def implication_elimination(self, sentence):
        # (P => Q) => (NOT P OR Q)
        if '=>' in sentence and '<=>' not in sentence:
            left, right = sentence.split('=>')
            return f'(NOT {left.strip()} OR {right.strip()})'
        return sentence

    def biconditional_elimination(self, sentence):
        # (P <=> Q) => (P => Q) AND (Q => P)
        if '<=>' in sentence:
            left, right = sentence.split('<=>')
            return f'({left.strip()} => {right.strip()}) AND ({right.strip()} => {left.strip()})'
        return sentence

    def double_negation(self, sentence):
        # Replace NOT NOT P with P
        return sentence.replace('NOT NOT ', '')

    def de_morgan(self, sentence):
        # NOT (A AND B) -> (NOT A OR NOT B), NOT (A OR B) -> (NOT A AND NOT B)
        import re
        s = sentence
        s = re.sub(r'NOT \(([^()]+) AND ([^()]+)\)', r'(NOT \1 OR NOT \2)', s)
        s = re.sub(r'NOT \(([^()]+) OR ([^()]+)\)', r'(NOT \1 AND NOT \2)', s)
        return s

    def distributive(self, sentence):
        # A OR (B AND C) -> (A OR B) AND (A OR C)
        import re
        s = sentence
        match = re.match(r'(.+) OR \((.+) AND (.+)\)', s)
        if match:
            a = match.group(1).strip()
            b = match.group(2).strip()
            c = match.group(3).strip()
            return f'({a} OR {b}) AND ({a} OR {c})'
        return s
```

## Step 3: Convert to CNF

Resolution requires all formulas to be in **Conjunctive Normal Form (CNF)**.

You will implement a pipeline to transform any formula into CNF.

```python
class CNFConverter:
    def to_cnf(self, sentence):
        # Step 1: Eliminate biconditionals
        engine = InferenceEngine()
        s = engine.biconditional_elimination(sentence)
        # Step 2: Eliminate implications
        if '<=>' not in s:
            s = engine.implication_elimination(s)
        # Step 3: Move NOT inward (De Morgan)
        s = engine.de_morgan(s)
        # Step 4: Double negation
        s = engine.double_negation(s)
        # Step 5: Distributive laws
        s = engine.distributive(s)
        return s
```

## Step 4: Resolution-Based Theorem Proving

Use the resolution rule to determine if a query is entailed.

This method uses **proof by contradiction**.

```python
class ResolutionProver:
    def resolve(self, clause1, clause2):
        # Find complementary literals (A and NOT A), remove them, combine rest
        c1 = set([x.strip() for x in clause1.replace('(', '').replace(')', '').split('OR')])
        c2 = set([x.strip() for x in clause2.replace('(', '').replace(')', '').split('OR')])
        for l1 in c1:
            if l1.startswith('NOT '):
                comp = l1[4:]
            else:
                comp = 'NOT ' + l1
            if comp in c2:
                new_clause = (c1 - {l1}) | (c2 - {comp})
                return ' OR '.join(sorted(new_clause))
        return None

    def prove(self, kb, query):
        # Convert KB to CNF, add NOT query, apply resolution
        cnf = []
        converter = CNFConverter()
        for s in kb.sentences:
            cs = converter.to_cnf(s)
            if 'AND' in cs:
                cnf.extend([c.strip() for c in cs.split('AND')])
            else:
                cnf.append(cs.strip())
        # Add NOT query
        not_query = f'NOT {query}'
        nq_cnf = converter.to_cnf(not_query)
        if 'AND' in nq_cnf:
            cnf.extend([c.strip() for c in nq_cnf.split('AND')])
        else:
            cnf.append(nq_cnf.strip())
        new = set()
        while True:
            n = len(cnf)
            pairs = [(i, j) for i in range(n) for j in range(i+1, n)]
            for (i, j) in pairs:
                resolvent = self.resolve(cnf[i], cnf[j])
                if resolvent is not None:
                    if resolvent == '':
                        return True  # Empty clause found
                    new.add(resolvent)
            if new.issubset(set(cnf)):
                return False
            for c in new:
                if c not in cnf:
                    cnf.append(c)
```

## Step 5: Model Checking

Model checking verifies if a query is true in **all possible models**.

```python
import itertools

def extract_symbols(sentences):
    symbols = set()
    for s in sentences:
        for token in s.replace('(', '').replace(')', '').replace('=>', ' ').replace('<=>', ' ').replace('AND', ' ').replace('OR', ' ').replace('NOT', ' ').split():
            if token.isalpha() and len(token) == 1:
                symbols.add(token)
    return sorted(symbols)

def eval_sentence(sentence, model):
    s = sentence
    for k, v in model.items():
        s = s.replace(f'NOT {k}', str(not v))
        s = s.replace(k, str(v))
    s = s.replace('AND', 'and').replace('OR', 'or')
    try:
        return eval(s)
    except:
        return False

def model_check(kb, query):
    symbols = extract_symbols(kb.sentences + [query])
    for values in itertools.product([False, True], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        kb_true = all(eval_sentence(s, model) for s in kb.sentences)
        if kb_true:
            if not eval_sentence(query, model):
                return False
    return True
```

## Step 6: File Input

Read the knowledge base and query from files.

```python
def read_kb(filename):
    kb = KnowledgeBase()
    with open(filename, 'r') as f:
        for line in f:
            line = line.strip()
            if line:
                kb.add(line)
    return kb

def read_query(filename):
    with open(filename, 'r') as f:
        return f.readline().strip()
```

## Step 7: Output Writer
Do not modify this function.

```python
def write_output(results, filename="output.txt"):
    output = "Inference Results:\n"
    output += "-------------------\n"

    for method, result in results.items():
        output += f"{method}: {result}\n"

    print(output)

    with open(filename, "w") as f:
        f.write(output)
```

## Step 8: Main Execution

Run both inference methods:

- Model Checking
- Resolution

Compare their results.

```python
def main():
    kb = read_kb("knowledge_base.txt")
    query = read_query("query.txt")

    results = {}

    # Run Model Checking
    results["Model Checking"] = model_check(kb, query)

    # Run Resolution
    prover = ResolutionProver()
    results["Resolution"] = prover.prove(kb, query)

    write_output(results)


if __name__ == "__main__":
    main()
```

## Step 10: Conceptual Questions

1. Why is CNF required for resolution?

2. Compare:
   - Model Checking
   - Resolution

3. What is proof by contradiction?

4. Why is model checking expensive?

5. How do inference rules improve reasoning efficiency?

```python
# Answers
# 1. CNF is required for resolution because the resolution rule operates on clauses in conjunctive normal form, allowing systematic application of the rule to pairs of clauses.
# 2. Model Checking checks all possible models for entailment, which is exhaustive but slow for large KBs. Resolution is a proof-based method that tries to derive a contradiction, often more efficient for large or infinite domains.
# 3. Proof by contradiction assumes the negation of the query and tries to derive a contradiction, thus proving the original query must be true.
# 4. Model checking is expensive because the number of possible models grows exponentially with the number of symbols.
# 5. Inference rules allow us to derive new facts efficiently without enumerating all models, making reasoning faster and more scalable.
```
