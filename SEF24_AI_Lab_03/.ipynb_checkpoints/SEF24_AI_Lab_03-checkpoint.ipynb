```python
import heapq


class Graph:
    def __init__(self):
        self.nodes = {}
        self.adjacency = {}

    def add_node(self, node_id, x, y):
        self.nodes[node_id] = (float(x), float(y))
        self.adjacency[node_id] = []

    def add_edge(self, node1, node2, cost):
        self.adjacency[node1].append((node2, float(cost)))
        self.adjacency[node2].append((node1, float(cost)))
```

##Step 4:

Run the output file writer. Do NOT modify this code.

```python
def write_output(path, cost, filename="output.txt"):
    result = "A* Search Results\n"
    result += "-----------------\n"

    if path:
        result += "Path: " + " -> ".join(path) + "\n"
        result += f"Total Cost: {round(cost, 2)}\n"
    else:
        result += "No path found\n"

    print(result)

    with open(filename, 'w') as file:
        file.write(result)
```
