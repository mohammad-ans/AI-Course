# Lab: First-Order Logic (FOL)

## Case Study: University Knowledge & Enrollment System

### Scenario

You are building a **University Knowledge Base System** using First-Order Logic.

The system models:

- Students
- Courses
- Enrollments
- Course categories (Core / Elective)

The goal is to:

- Represent facts using predicates
- Evaluate logical statements
- Apply Universal (∀) and Existential (∃) quantifiers
- Perform reasoning on relationships

---

### Predicates in the System

- Student(s)
- Course(c)
- Enrolled(s, c)
- CoreCourse(c)

---

### Example Logical Statements

1. Student(Ali)  
2. Course(CS101)  
3. Enrolled(Ali, CS101)  
4. CoreCourse(CS101)

---

### Quantified Statements

1. ∀s (Enrolled(s, CS101) → Student(s))  
2. ∃c (Course(c) ∧ Enrolled(Ali, c))  
3. ∀s (Student(s) → ∃c Enrolled(s, c))  
4. ∃s ∀c (Course(c) → ¬Enrolled(s, c))

## Step 1

Define the Knowledge Base class.

```python
class UniversityKB:
    def __init__(self):
        self.students = []
        self.courses = []
        self.enrollments = set()
        self.core_courses = set()
        self.completed = set()

    # Add methods
    def add_student(self, s):
        if s not in self.students:
            self.students.append(s)

    def add_course(self, c):
        if c not in self.courses:
            self.courses.append(c)

    def add_enrollment(self, s, c):
        self.enrollments.add((s, c))

    def add_core_course(self, c):
        self.core_courses.add(c)

    def add_completed(self, s, c):
        self.completed.add((s, c))

    # Predicates
    def Student(self, s):
        return s in self.students

    def Course(self, c):
        return c in self.courses

    def Enrolled(self, s, c):
        return (s, c) in self.enrollments

    def CoreCourse(self, c):
        return c in self.core_courses

    def Completed(self, s, c):
        return (s, c) in self.completed
```

## Step 2

Complete file reading functions.

Hint:
- Use file reading
- Strip newline characters
- Split where needed

```python
def read_students(file, kb):
    with open(file, 'r') as f:
        for line in f:
            kb.add_student(line.strip())


def read_courses(file, kb):
    with open(file, 'r') as f:
        for line in f:
            kb.add_course(line.strip())


def read_enrollments(file, kb):
    with open(file, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 2:
                kb.add_enrollment(parts[0], parts[1])


def read_core_courses(file, kb):
    with open(file, 'r') as f:
        for line in f:
            kb.add_core_course(line.strip())
```

## Step 3

Implement First-Order Logic Evaluations.

Use loops to simulate quantifiers:

- ∀ → check ALL elements
- ∃ → check if ANY element satisfies condition

```python
class FOLEvaluator:
    def __init__(self, kb):
        self.kb = kb

    def universal_cs101_students(self):
        """
        ∀s (Enrolled(s, CS101) → Student(s))
        """
        for s in self.kb.students:
            if self.kb.Enrolled(s, 'CS101') and not self.kb.Student(s):
                return False
        return True

    def existential_ali_enrolled(self):
        """
        ∃c (Course(c) ∧ Enrolled(Ali, c))
        """
        for c in self.kb.courses:
            if self.kb.Course(c) and self.kb.Enrolled('Ali', c):
                return True
        return False

    def every_student_has_course(self):
        """
        ∀s (Student(s) → ∃c Enrolled(s, c))
        """
        for s in self.kb.students:
            if self.kb.Student(s):
                enrolled = False
                for c in self.kb.courses:
                    if self.kb.Enrolled(s, c):
                        enrolled = True
                        break
                if not enrolled:
                    return False
        return True

    def exists_student_with_no_courses(self):
        """
        ∃s ∀c (Course(c) → ¬Enrolled(s, c))
        """
        for s in self.kb.students:
            no_enrollment = True
            for c in self.kb.courses:
                if self.kb.Enrolled(s, c):
                    no_enrollment = False
                    break
            if no_enrollment:
                return True
        return False

    def core_course_check(self):
        """
        ∀c (CoreCourse(c) → ∃s Enrolled(s, c))
        """
        for c in self.kb.core_courses:
            enrolled = False
            for s in self.kb.students:
                if self.kb.Enrolled(s, c):
                    enrolled = True
                    break
            if not enrolled:
                return False
        return True

    def every_student_completed_course(self):
        """
        ∀s (Student(s) → ∃c Completed(s, c))
        """
        for s in self.kb.students:
            if self.kb.Student(s):
                completed = False
                for c in self.kb.courses:
                    if self.kb.Completed(s, c):
                        completed = True
                        break
                if not completed:
                    return False
        return True

    def students_with_multiple_enrollments(self):
        """
        Find students enrolled in more than one course
        """
        result = []
        for s in self.kb.students:
            count = 0
            for c in self.kb.courses:
                if self.kb.Enrolled(s, c):
                    count += 1
            if count > 1:
                result.append(s)
        return result

    def every_student_enrolled_in_core(self):
        """
        ∀s ∃c (CoreCourse(c) ∧ Enrolled(s, c))
        """
        for s in self.kb.students:
            if self.kb.Student(s):
                enrolled_in_core = False
                for c in self.kb.core_courses:
                    if self.kb.Enrolled(s, c):
                        enrolled_in_core = True
                        break
                if not enrolled_in_core:
                    return False
        return True
```

## Step 4

Complete evaluation function.

```python
def evaluate(kb):
    evaluator = FOLEvaluator(kb)

    results = []

    results.append(f"Universal CS101 Students: {evaluator.universal_cs101_students()}")
    results.append(f"Existential Ali Enrolled: {evaluator.existential_ali_enrolled()}")
    results.append(f"Every Student Has Course: {evaluator.every_student_has_course()}")
    results.append(f"Exists Student With No Courses: {evaluator.exists_student_with_no_courses()}")
    results.append(f"Core Course Check: {evaluator.core_course_check()}")
    results.append(f"Exists Student with Multiple Enrollments: {evaluator.students_with_multiple_enrollments()}")
    results.append(f"Every Student completed course: {evaluator.every_student_completed_course()}")
    
    results.append(f"Every Student enrolled in a core course: {evaluator.every_student_enrolled_in_core()}")

    return results
```

## Step 5

Do NOT modify this code.

```python
def write_output(results, filename="output.txt"):
    output = "FOL Evaluation Results\n"
    output += "----------------------\n"

    for r in results:
        output += r + "\n"

    print(output)

    with open(filename, "w") as f:
        f.write(output)
```

## Step 6

Main function.
Do NOT modify.

```python
def main():
    kb = UniversityKB()

    read_students("students.txt", kb)
    read_courses("courses.txt", kb)
    read_enrollments("enrollments.txt", kb)
    read_core_courses("core_courses.txt", kb)

    results = evaluate(kb)
    write_output(results)


if __name__ == "__main__":
    main()
```
