## Step 1: Load the Graph

We will read graph.txt and convert it into an adjacency list.

Expected structure:

{
    'A': [('B', 4), ('C', 3)],
    'B': [('D', 5)],
    ...
}

```python
def load_graph(filename):
    graph = {}

    with open(filename, 'r') as file:
        for line in file:
            parts = line.strip().split()

            if len(parts) != 3:
                continue

            src, dest, cost = parts
            cost = int(cost)
            if src not in graph:
                graph[src] = ([(dest, cost)])
                
            else:
                graph[src].append((dest,cost))
            
            # TODO: Populate adjacency list
    return graph
```

## Step 2: Load Heuristic Values

The heuristic function estimates the distance from a node to the goal.

Example structure:

```python
def load_heuristic(filename):
    """
    Reads heuristic values from file.
    The file can be read the same way as it is being read in Step 1.
    Returns dictionary:
    {
        'A': 10,
        'B': 8,
        ...
    }
    """
    heuristic = {}
    with open(filename, 'r') as file:
        for line in file:
            parts = line.strip().split()

            if len(parts) != 2:
                continue

            src, h = parts
            h = int(h)
            heuristic[src] = h
    # TODO: Open file
    # TODO: Read node and heuristic value
    # TODO: Store in dictionary

    return heuristic
```

## Step 3: Implement Greedy Best-First Search

Algorithm:

1. Insert start node into priority queue with priority h(start)
2. While frontier is not empty:
   - Remove node with smallest h(n)
   - If node is goal, stop
   - Mark node as visited
   - Add unvisited neighbors to priority queue
3. Reconstruct path

```python
def greedy_best_first_search(graph, heuristic, start, goal):
    """
    Implements GBFS algorithm.

    Returns:
        path (list)
        total_cost (int)
        explored_order (list)
    """

    # Priority queue: (heuristic_value, node)
    frontier = []
    # TODO: Push start node into priority queue
    heapq.heappush(frontier, (heuristic[start], start))
    visited = set()

    parent = {}  # To reconstruct path
    explored_order = []
    while frontier:

        # TODO: Pop node with lowest heuristic value
        _, current = frontier.pop()
        # TODO: Add node to explored_order
        explored_order.append(current)
        # TODO: Check if goal reached
        if current == goal:
            path = reconstruct_path(parent, start, goal)
            cost = calculate_cost(graph, path)
            return path, cost, explored_order
        # TODO: Mark node as visited
        visited.add(current)
        # TODO: Expand neighbors
        # For each neighbor:
        #   If not visited:
        #       Set parent
        #       Push into priority queue with heuristic value
        for element in graph[current]:
            neighbour = element[0]
            if not neighbour in visited:
                parent[neighbour] = current
                heapq.heappush(frontier, (heuristic[neighbour], neighbour))
                

    # TODO: If goal not found, return None

    return None, None, explored_order
```

## Step 4: Reconstruct the Path

We backtrack from the goal to the start using the parent dictionary.

```python
def reconstruct_path(parent, start, goal):
    """
    Reconstructs path from start to goal.
    """

    path = []
    while goal in parent:
        path.append(goal)
        goal = parent[goal]
    path.append(start)
    return path[::-1]
    # TODO: Backtrack from goal to start using parent dictionary

    return path
```

## Step 5: Calculate Path Cost

Although GBFS ignores cost during search,
we compute the actual cost afterward for evaluation.

```python
def calculate_cost(graph, path):
    """
    Calculates total cost of path.
    """

    total_cost = 0

    # TODO: Loop through path and sum edge costs
    for index, element in enumerate(path):
        if index < len(path) - 1:
            temp = path[index + 1]
            for current in graph[element]:
                if current[0] == temp:
                    total_cost += current[1]
                    break
        
    return total_cost
calculate_cost({'A': [('B', 4), ('C', 3)],
 'B': [('D', 5), ('E', 12)],
 'C': [('F', 7)],
 'F': [('G', 2)],
 'D': [('G', 9)],
 'E': [('G', 1)]}, ['A', 'C', 'F'])
```
