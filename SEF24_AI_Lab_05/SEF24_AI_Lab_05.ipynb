## Case Study: Smart Home Rule Engine with Complex Conditions

### Scenario

You are building a **smart home rule engine**. The system monitors sensors:

- motion detected
- door open
- temperature high
- smoke detected
- time of day

Each rule is a **complex logical expression** that may include:

- parentheses `()`
- operators: `AND`, `OR`, `NOT`
- multiple sensor references

Example rules:

- `(motion AND door) OR time_night`
- `NOT smoke AND (temperature OR motion)`

Your task:

- Parse infix logical formulas
- Evaluate them given current sensor inputs
- Determine which rules are triggered

## Input Files

### sensors.txt

sensor_name True/False

Example:

motion True
door False
temperature True
smoke False
time_night True

---

### rules.txt

rule_name: formula

Formulas may include parentheses and multiple operators.

Example:

rule1: motion AND door
rule2: temperature OR smoke
rule3: NOT time_night OR door
rule4: motion AND (NOT door OR time_night)

## Step 1

Review the classes used to represent sensors and rules.

```python
class Sensor:
    def __init__(self, name, value):
        self.name = name
        self.value = value


class Rule:
    def __init__(self, name, formula):
        self.name = name
        self.formula = formula 

    def evaluate(self, sensor_dict):
        exp = self.formula

        for s in sensor_dict:
            exp = exp.replace(s, str(sensor_dict[s]))

        exp = exp.replace("AND", "and")
        exp = exp.replace("OR", "or")
        exp = exp.replace("NOT", "not")

        return eval(exp)
```

## Step 2

Implement file reader functions.

```python
def read_sensors(filename):
    sensor_dict = {}
    
    with open(filename, "r") as file:
        
        for line in file:
            parts = line.strip().split()
            name = parts[0]
            value = parts[1] == "True"
            sensor = Sensor(name, value)
            sensor_dict[name] = sensor.value
            
    return sensor_dict


def read_rules(filename):
    rules = []
    
    with open(filename, "r") as file:
        
        for line in file:
            name, formula = line.strip().split(":")
            rule = Rule(name.strip(), formula.strip())
            rules.append(rule)
            
    return rules
```

## Step 3

Implement the main evaluation loop.

```python
def evaluate_rules(sensor_dict, rules):
    triggered_rules = []
    for rule in rules:
         if rule.evaluate(sensor_dict):
            triggered_rules.append(rule.name)
            
    return triggered_rules
```

## Step 4

Output the triggered rules. Do NOT modify this code.

```python
def write_output(triggered_rules, filename="output.txt"):
    result = "Triggered Rules:\n"
    result += "----------------\n"
    for rule in triggered_rules:
        result += rule + "\n"

    print(result)

    with open(filename, "w") as file:
        file.write(result)
```

## Step 5

Run the main function. Do NOT modify this code.

```python
def main():
    sensor_dict = read_sensors("sensors.txt")
    rules = read_rules("rules.txt")
    triggered_rules = evaluate_rules(sensor_dict, rules)
    write_output(triggered_rules)


if __name__ == "__main__":
    main()
```
